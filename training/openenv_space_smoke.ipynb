# GOAT-Evolution HF Space OpenEnv Smoke Test

Quick notebook to verify your deployed Space is serving OpenEnv endpoints.

```python
import requests, json

SPACE_BASE = "https://YOUR_USER-YOUR_SPACE.hf.space"  # set your runtime URL
print("Testing:", SPACE_BASE)
```

```python
for path in ["/health", "/metadata", "/schema"]:
    r = requests.get(SPACE_BASE + path, timeout=30)
    print(path, r.status_code)
    print(r.text[:300], "\n")
```

```python
reset_payload = {
    "task_id": 4,
    "episode_id": "hf-smoke",
    "model_profile": "standard_model",
}
reset_resp = requests.post(SPACE_BASE + "/reset", json=reset_payload, timeout=30)
print("/reset", reset_resp.status_code)
reset_json = reset_resp.json()
print(reset_json.keys())
print("done=", reset_json.get("done"), "reward=", reset_json.get("reward"))
```

```python
step_payload = {
    "action": {
        "role": "red",
        "action_json": json.dumps({
            "role": "red",
            "red_payload": {
                "operation": "deploy_strategy",
                "current_strategy": ["crescendo"],
                "thought": "smoke test",
            },
        }),
    }
}
step_resp = requests.post(SPACE_BASE + "/step", json=step_payload, timeout=30)
print("/step", step_resp.status_code)
step_json = step_resp.json()
print("done=", step_json.get("done"), "reward=", step_json.get("reward"))
print("observation keys:", list(step_json.get("observation", {}).keys())[:12])
```
